```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
from wordcloud import WordCloud

plt.rcParams.update(
    {"figure.figsize": (10, 5), "figure.dpi": 200, "figure.max_open_warning": 0}
)
plt.style.use("dark_background")
```

```python
df = pd.read_excel('../static/output/participacao_em_bancas.xlsx')
```

```python
print(f"Dataframe contém {df.shape[0]} linhas e {df.shape[1]} colunas.")
```

```python
print("Columns in the dataframe: ", ", ".join(df.columns))
```

# Participações em bancas por ano

Este gráfico permite identificar a evolução do docente em termos de participações em bancas ao longo dos anos.
Demonstrando o envolvimento acadêmico na produção científica e na formação de novos pesquisadores. Essa análise é importante para entender o impacto do docente na comunidade acadêmica.


```python
# Hist plot of the number of participations per year
# grouped by `uuid` to count unique participations
yearly_counts = df['ano'].groupby(df['uuid']).min().value_counts().sort_index()
fig, ax = plt.subplots()
ax.hist(yearly_counts.index, weights=yearly_counts.values, 
        bins=np.arange(yearly_counts.index.min(), 
        yearly_counts.index.max()+2)-0.5, edgecolor='black',
        label='Participações Únicas', histtype='bar', alpha=0.7) 
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
ax.set_xticks(np.arange(yearly_counts.index.min(), yearly_counts.index.max()+1, step=1))
ax.set_yticks(np.arange(0, yearly_counts.values.max()+1, step=5))
ax.grid(axis='y', linestyle='--', alpha=0.7)
ax.set_xlabel('Ano')
ax.set_ylabel('Número de Participações Únicas')
ax.set_title('Número de Participações Únicas em Bancas por Ano')
plt.tight_layout()
plt.show()
```

# Gráfico de setores da distribuição dos tipos de participações
Este gráfico de setores ilustra a distribuição dos diferentes tipos de participações em bancas acadêmicas.
Ele ajuda na visualização da proporção de cada participação, permitindo compreender melhor o as bancas de maior participação do docente.

```python
# Sector plot of the distribution of types of participations
type_counts = df['natureza'].groupby(df['uuid']).min().value_counts()
fig, ax = plt.subplots()
ax.pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%', startangle=140)
ax.set_title('Distribuição dos Tipos de Participações em Bancas')
ax.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
fig.tight_layout()
plt.show()
```

# Nuvem de palavras chave
A nuvem de palavras-chave oferece uma visão geral das principais áreas de interesse e pesquisa do docente.
As palavras mais frequentes aparecem em destaque, facilitando a identificação dos temas centrais abordados em suas participações acadêmicas.

```python

# Wordcloud of Keywords

phrase_counts = Counter(df['palavra_chave'].dropna().groupby(df['uuid']).min())
wordcloud = WordCloud(width=1920, height=1080, background_color='black').fit_words(phrase_counts)
fig, ax = plt.subplots(figsize=(10, 5))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
ax.set_title('Nuvem de Palavras-chave das Participações em Bancas')
plt.tight_layout()
plt.show()
```
